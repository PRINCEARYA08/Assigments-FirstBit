```python
# serching methods

str="Good in Python Good in"
print(str.find("in"))  # for maching elemint
print(str.find("om"))  # -1 for not maching elemint
print(str.find("python"))
print(str.rfind("Good")) # it returns last matching index otherwise return -1
print(str.index("in"))
print(str.rindex("Good"))
print(str.count("Good")) # it will print the count of matching string if not found then it retun -1
```
