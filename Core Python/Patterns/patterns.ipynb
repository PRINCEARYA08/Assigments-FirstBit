```python
#1
for i in range(1,6):
    for j in range(1,6):
        print(i,end=" ")
    print()    #print() in python it give next line by default
    
# similar value in row: use i
```

```python
#2
for i in range(1,6):
    for j in range(1,6):
        print(j,end=" ")
    print()
    
    # varies value in row: use j
```

```python
#3
for i in range(1,6):
   
   for j in range(i):
       print("*", end=" ")
   print()
       
```
