```python
for i in range(1,6):
    for j in range(1,6):
        print(i,end=" ")
    print()
    
# similar value in row: use i
#p1
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(j,end=" ")
    print()
    
    # varies value in row: use j
#p2
```

```python
for i in range(1,6):
   
   for j in range(i):
       print("*", end=" ")
   print()
       
#3
```
