## lambda
### 1. reduce lines of code
### 2. anonymous function
### 3. one liner function
### 4. one exppression function
### 5. always return a value or output


```python
# lambda parameters: expression

add = lambda num1, num2: num1 + num2

res = add(10, 20)
print(res)

```

```python
opr = lambda num1, num2: num1 * num2

output = opr(10, 20)
print(output)
```

```python
p = float(input("Enter a number:"))
t = float(input("Enter time: "))
r = float(input("Enter rate: "))

si = lambda p,t,r: (p*t*r)/100
print("Simple Interest is: ", si(p,t,r))
```

# Map function
### 1. to reduce line of code
### 2. perform same task multiple times on diff input


```python
#method 1

def sq(n):
    return n*n

data = [1,2,3,4,5,6,7,8,9]


res = list(map(sq, data))
print(res)
```

```python
#method2
data = [1,2,3,4,5,6,7,8,9]


res = list(map(lambda n: n*n, data))
print(res)
```

```python
## filter

data = [1,2,3,4,5,6,7,8,9]

res =list(filter(lambda n: n*n, data))
print(res)
```

```python
## filter


# to print even numbers from the list
data = [1,2,3,4,5,6,7,8,9]

res =list(filter(lambda n: n%2==0, data))
print(res)
```

```python
## filter


# to print odd numbers from the list
data = [1,2,3,4,5,6,7,8,9]

res = list(filter(lambda n: n % 2 != 0, data))
print(res)
```

```python
from functools import reduce 
data = [1,2,3,4,5,6,7,8,9]

res = reduce(lambda x, y: x+y, data)
print(res)
```
