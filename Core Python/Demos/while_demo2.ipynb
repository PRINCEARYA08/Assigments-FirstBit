```python
num = 4927
sum = 0
while(num>0):     # here 0 is the condition for the loop to run, as long as num is greater than 0, the loop will continue to execute
    d = num % 10
    print(d)
    num = num//10
    sum+=d
    
print("Sum of digits:",sum)
```
