#### 1. Numeric Data Type

```python

#1. int 
#int var # variable declaration
var = 10   # variable initialization
print(var, type(var))
```

```python
#2. float
var = 10.5
print(var, type(var))
```

```python
#3 complex
var = 10 + 5j  #real + imaginary
print(var, type(var))
```

#### 2. Text

```python
# str
var1 = "Hello1"
var2 = 'Hello2'
var3 = """Hello
     MULTILINE STRING  3"""
print(var1, type(var1))
print(var2, type(var2))
print(var3, type(var3))
```

#### 3. Sequential


```python
#1.list
var = [10, 20, 30, 40]
print(var, type(var))
```

```python
# tuple
var = (10, 20, 30, 40)
print(var, type(var))
```

```python
# range
var = range(1, 10)

print(var)
print(var, type(var))
print(list(var))  #converting range to list

```

#### 4. set type

```python
#1. set
var = {10, 20, 30, 40}
var1 = {10, 20, 30, 40, 10, 20} #duplicate values will be removed

print(var, type(var))
print(var1, type(var1))
```

```python
# frozenset
var = frozenset([10, 20, 30, 40])
print(var, type(var))
```
