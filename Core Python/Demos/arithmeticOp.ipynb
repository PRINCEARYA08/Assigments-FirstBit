```python
x = 10
y = 20
a = "abc"
b = "cde"

#1.  +
#addition of two numbers
print(x + y)  # 30

#2.  - 
print(x - y)  # -10

# concatination of two strings
print(a + " " + b)  # abc cde

# res = x + a
# print(res)  # TypeError: unsupported operand type(s) for +: 'int' and 'str'

#3. *
print(x * y)  # 200

#4. /  (floor division operator)
print(5 / 2)  # 2.5    count after decimal point

#5. // 
print(5 // 2)  # 2   count before decimal point

#6. %   (modulus operator)
print(x % y)  # 10    remainder of division


#7. **  (exponentiation operator)
print(8** 8)  # 100    x raised to the power of 2
```
