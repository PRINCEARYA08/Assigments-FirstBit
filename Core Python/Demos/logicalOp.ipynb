```python
# and : if both operands are true
# then return true otherwise return false

print(True and True)   # True
print(True and False)  # False
print(False and True)  # False
print(False and False) # False

 
```

```python
#2 or : if any one operand is true then return true otherwise return false
# if one condition is true then it will not check the second condition return true 


print(True or True)   # True
print(True or False)  # True
print(False or True)  # True
print(False or False) # False
```

```python
# reserch on how it work
print('' or 10)
```

```python
#3 not : if the operand is true then return false otherwise return true
# opposite of the operand
print(not True)   # False
print(not False)  # True
```
