```python
# pass : neglect the expected indentation error and continue the execution of the program

for i in range(5):
    pass
```

```python
# breack : break the loop and come out of the loop
# use to terminate the loop when a certain condition is met
for i in range(5):
    if i==3:      # when i is equal to 3 then break the loop 
        break
    print(i)
```

```python
# continue : skip the current iteration of the loop and continue with the next iteration
#
for i in range(5):
    if i==3:      # when i is equal to 3 then skip the current iteration and continue with the next iteration
        continue  # it skip 3 and jump to 4
    print(i)
```

```python
# else:  
for i in range(1,5):
    if(i==6):
        break
    print(i)
else:
    print("else bloack is exicuted.")  
    # it will exicute because the loop is not terminated by break statement.
    # if the loop is terminated by break statement then else block will not be exicuted. 
```
