```python
# this program calculates the sum of first n numbers
num= int(input("Enter a number "))
sum=0
for i in range(1,num+1):
    #print(i)
    sum+=i
print("The sum of first",num,"numbers is",sum)
```

```python
# this program calculate the factorial of a number   
# 5!
num= int(input("Enter a number "))
sum=1
for i in range(1,num+1):
    #print(i)
    sum*=i
print("The sum of first",num,"numbers is",sum)
```
